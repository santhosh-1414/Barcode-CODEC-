```python
# importing required libraries.
import cv2
import numpy as np

# class Barcode has the encode and decode functions responsible for creating and decoding a barcode.
class Barcode:

    # encode function to draw a barcode of the given string.
    def encode(self, input_text):
        # blank white screen of width 800 pixels and height 400 pixels.
        canvas = np.ones((400, 800), dtype=np.uint8) * 255
        # alphabets variable to calculate the width of bars based on the position in this variable.
        alphabets = 'abcdefghijklmnopqrstuvwxyz'
        # Step is space between each bar 
        step = 9  
        # strating position.
        x = 0
        # for each letter in the input (converted to small case)
        for char in input_text.lower():
            if char == ' ':
                # if space is there make width 1 and height short height (150, 250)
                bar_x = 1
                bar_y = (150, 250)
            elif char in alphabets:
                # if it is a letter then width is add +2 to the position of that letter in the alphabet variable
                bar_x = alphabets.index(char) + 2
                # height is full height (10, 350)
                bar_y = (10, 350)
            else:
                # skip other characters meaning it wont draw bars for symbols and numbers
                continue

            # Drawing the bars for all the letters and spaces in the input
            for y in range(bar_y[0], bar_y[1]):
                canvas[y, x:x + bar_x] = 0 
            # go to the next position to draw
            x = x + bar_x + step

        # saving the barcode image as output.png
        cv2.imwrite('output.png', canvas)
        print("Barcode image is saved as output.png")

    def decode(self, ipath):
        # converting the image to black and white and loding it in read variable.
        read = cv2.imread(ipath, cv2.IMREAD_GRAYSCALE)
        # reading from the middle row of the image
        m_row = read[200, :]
        
        # initially the result string is empty.
        result = ''
        count = 0
        # alphabets variable to calculate the width of bars based on the position in this variable.
        alphabets = 'abcdefghijklmnopqrstuvwxyz'

        # foe every pixel from left to right in middle row
        for pixel in m_row:
            if pixel == 0:
                # If the pixel is 0 meaning it is black, increase the counter
                count += 1
            else:
                # if it is a white pixel and the count is one means it is a space
                if count == 1:
                    result = result + ' '  
                elif count > 1:
                    # if the count is more than 1 means it is a letter and removing 2 from count gives the index of the letter in the alphabets vaariable
                    result = result + alphabets[count - 2]
                # reseting for the next bar
                count = 0

        # printing the deocded string
        print("Decoded string from the input image is:")
        print(result)
        return result

# Example usage
barcode = Barcode()
barcode.encode('Abbas Cheddad')
barcode.decode('output.png')
```
